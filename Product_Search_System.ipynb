```python
import re

products = [
    "Apple iPhone",
    "Apple MacBook",
    "Samsung Phone",
    "Gaming Laptop",
    "Dell Laptop",
    "Wireless Mouse",
    "Gaming Mouse"
]

def search_products(keyword, search_type):
    if search_type == "exact":
        pattern = r"^" + re.escape(keyword) + r"$"

    elif search_type == "prefix":
        pattern = r"^" + re.escape(keyword)

    elif search_type == "suffix":
        pattern = re.escape(keyword) + r"$"

    elif search_type == "partial":
        pattern = re.escape(keyword)

    else:
        print("Invalid search type")
        return

    matches = []

    for product in products:
        if re.search(pattern, product, re.IGNORECASE):
            matches.append(product)

    print("\nSearch Type:", search_type)
    print("Keyword:", keyword)
    print("Matching Products:")

    for product in matches:
        print("-", product)

    print("Total Matches:", len(matches))


search_products("Apple", "prefix")
search_products("Laptop", "suffix")
search_products("Mouse", "partial")
```
