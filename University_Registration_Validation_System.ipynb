```python
import re

def validate_registration(reg_no, email, course, semester, mobile):

    # Validation patterns
    reg_pattern = r"^RA\d{13}$"
    email_pattern = r"^[A-Za-z0-9._%+-]+@saveetha\.com$"
    course_pattern = r"^[A-Z]{3}\d{2}$"
    semester_pattern = r"^[1-8]$"
    mobile_pattern = r"^[6-9]\d{9}$"

    validations = {
        "Register Number": bool(re.fullmatch(reg_pattern, reg_no)),
        "Email": bool(re.fullmatch(email_pattern, email)),
        "Course Code": bool(re.fullmatch(course_pattern, course)),
        "Semester": bool(re.fullmatch(semester_pattern, semester)),
        "Mobile Number": bool(re.fullmatch(mobile_pattern, mobile))
    }

    print("----- Validation Report -----")

    for field, status in validations.items():
        if status:
            print(field, ": Valid")
        else:
            print(field, ": Invalid")

    print("\n----- Final Registration Status -----")

    if all(validations.values()):
        print("Registration Successful")
    else:
        print("Registration Failed")


validate_registration(
    "RA2311003010001",
    "student@saveetha.com",
    "DSA03",
    "5",
    "9876543210"
)
```
